```python
# 1. Pedir ao usuário para digitar uma lista aninhada (ex: "1,2,3;4,5;6")
nested_input = input("Enter nested list (use commas and semicolons): ")

# 2. Converter a entrada em uma lista aninhada de números inteiros
nested_list = [[int(num) for num in sublist.split(",")] for sublist in nested_input.split(";")]

max = nested_list[0][0]

for sublist in nested_list:
    for numbers in sublist:
        if numbers > max: 
            max = numbers
            print(max)
```

```python
base = int(input('Insira a base do triângulo (use números ímpares para alinhar melhor): '))
i = 1

while i <= base:
    # 1. Calculate the necessary spaces to center the line
    espacos = (base - i) // 2
    
    # 2. Print the spaces followed by the asterisks
    print(' ' * espacos + '*' * i)
    
    # 3. Skip to the next odd row (1, 3, 5...)
    i = i + 2
```

```python
llista = [1, 2, 3, 4]

def produto_iterativo(lista):
    produto = 1  # Inicializa o produto com 1
    for elemento in lista:
        produto *= elemento  # Multiplica cada elemento ao produto
    return produto

# Chamando a função
resultado = produto_iterativo(lista)
print("Produto dos elementos:", resultado)  # Saída: 24 (1 * 2 * 3 * 4)

```

```python
# Função para verificar se o input é um número válido
def verificar_numero(mensagem):
    while True:
        try:
            numero = float(input(mensagem))
            return numero
        except ValueError:
            print("Erro: Por favor, digite um número válido.")

# Receber os dois números do usuário
numero1 = verificar_numero("Digite o primeiro número: ")
numero2 = verificar_numero("Digite o segundo número: ")

# Verificar se o segundo número não é zero para evitar divisão por zero
if numero2 == 0:
    print("Erro: Divisão por zero não é permitida.")
else:
    # Realizar as operações
    soma = numero1 + numero2
    multiplicacao = numero1 * numero2
    subtracao = numero1 - numero2
    divisao = numero1 / numero2

    # Imprimir os resultados
    print("\nResultados das operações:")
    print(f"Soma: {numero1} + {numero2} = {soma}")
    print(f"Multiplicação: {numero1} * {numero2} = {multiplicacao}")
    print(f"Subtração: {numero1} - {numero2} = {subtracao}")
    print(f"Divisão: {numero1} / {numero2} = {divisao:.2f}")
```
